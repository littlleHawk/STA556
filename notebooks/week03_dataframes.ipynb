```python
import pandas as pd
```

```python
data = {
    "id": [101, 102, 103, 104, 105, 106],
    "age": [24, 31, 27, 19, 42, 35],
    "group": ["A", "B", "A", "B", "A", "B"],
    "score": [81, 94, 88, 72, 91, 85]
}

df = pd.DataFrame(data)
```

```python
print(f"{df=}")
print(f"{df.head()=}")
print(f"{df.shape=}")
print(f"{df.columns=}")
print(f"{df.index=}")
print(f"{df.dtypes=}")
print(f"{df.info()=}")
print(f"{df.describe()=}")
```

```python
age = df["age"]
type(age)
```

```python
age_df = df[["age"]]
type(age_df)
```

df["age"] selects just the column as a pd.Series, while df[["age"]] returns the vector as a dataframe datatype.

```python
analysis_df = df[["id", "group", "score"]]
analysis_df
```

```python
print(f"\n{df.iloc[0]=}")
print(f"\n{df.iloc[0:3]=}")
print(f"\n{df.iloc[[0, 2, 5]]=}")
print(f"\n{df.iloc[1:4, 1:3]=}")
```

The first argumnet to `iloc()` controls the index of the rows selected, the second controls the columns. The stop value is not included in positonal slicing.

```python
df = df.set_index("id")
print(f"{df= }")
print(f"{df.index= }")
```

```python
print(f"{df.loc[101] = }")
# df.loc[0] is no longer available because the ID 0 does not exist. 
# They both should indicate the first line, but the ID of the first line is no longer 0

# Using iloc works with 0 becuase this is the index value, not the ID value
print(f"{df.iloc[0] = }")
```

```python
df.loc[[102, 104, 106], ["age", "group", "score"]]
```

```python
small = pd.DataFrame(
    {"value": [10, 20, 30, 40]},
    index=[101, 205, 310, 412]
)

print(f"{small.loc[205] = }")
print(f"{small.iloc[1] = }")
```

loc uses a named index value in the df, iloc uses positional indexing.

```python
df.iloc[1:4]
```

```python
df.loc[102:105]
```

Positional slicing contains up to the last position (not inclusive) while label-based slicing selects all that match the given labels. It would not make sense to make label slicing not inclusive because then we would need to know the label of the first non-included value.

```python
df.loc[:, ["age", "score"]]
```

```python
df.loc[
    101:104,
    ["group", "score"]
]
```

```python
condition = df["score"] > 80
```

```python
print(f"{condition = }")
print(f"{type(condition) = }")
```

True represents a score value larger than 80; false represents a score value of less than or equal to 80. 

```python
#younger than 30;
print(f"{df.loc[df["age"] < 30] = } ")

#in group B;
print(f"{df.loc[df["group"] == "B"] = } ")

#scoring at least 85.
print(f"{df.loc[df["score"] >= 85] = } ")
```

```python
#Find participants who are at least 30 years old and score at least 85.
df.loc[(df["age"]>=30)&(df["score"]>=85)]
```

Score = 0 is where the value of the observation was actially 0. Score = missing is where no value was observed at all. In statistical analysis, a 0 might be real value that affects the modelling and summary of the analysis. Missing values will not pull summaries or models in any direction.

# Mini data science challenge

```python
participants = pd.DataFrame(
    {
        "id": range(1001, 1011),
        "age": [22, 34, 29, 41, 19, 37, 26, 45, 31, 28],
        "group": ["A", "B", "A", "C", "B", "A", "C", "B", "A", "C"],
        "score": [78, 91, 85, 88, 72, 95, 81, 67, 90, 84]
    }
).set_index("id")
```

```python
# A: 30 or older
older = participants.loc[participants["age"] >= 30]
# B: Groups a or c
groups_ac = participants.loc[participants["group"].isin(["A", "C"])]
# C: scores >= 85
high_scores = participants.loc[participants["score"] >= 85]
# D: at least 30 yo and 85 score
older_high = participants.loc[(participants["age"] >= 30) & (participants["score"]>= 85)]
# E: Group and score for high scorers
high_score_summary = participants.loc[(participants["score"] >= 85), ["group", "score"]]
# F: Group A and at least score 80 
group_a_good = participants.loc[(participants["group"] == "A")&(participants["score"] >= 80)]
# G: Not group B
not_group_b = participants.loc[participants["group"] != "B"]
```

```python
dfs = [older,
groups_ac,
high_scores,
older_high,
high_score_summary,
group_a_good,
not_group_b]
```

```python
for df in dfs:
    print(f"\n===================================\n")
    print(f"\n{df.shape = }")
    print(f"\n{df.head() = }")
    print(f"\n{df.index = }")
    print(f"\n")
```
