# Everything is an Object

```python
x = 42

print(f"{type(x)=}")
print(f"{id(x)=}")
print(f"{repr(x)=}")
```

```python
x = "hello"

print(f"{type(x)=}")
print(f"{id(x)=}")
print(f"{repr(x)=}")
```

1. Did the type change?
* Yes, the type changed from `class 'int'` to `class 'str'` because 42 and "hello" are different object typed
2. Did the identity change?
* The identity did change, x originally pointed at one spot in memory, then pointed somewhere new when the value changed.
3. What happened to the integer object?
* The integer object likely still exists at that place in memory, but is no longer accessible because nothing points to it anymore,
4. Does the name x have a permanent type?
* No (yes, object but class type no)

## Part 2: Names and Objects

```python
x = [1, 2, 3]
y = x

x is y
```

```python
x.append(4)

print(x)
# Y will have the same value as x [1, 2, 3, 4] because both point to the same list, which gets mutated in-place
# a new space in memory is not created for the mutated list because python lists are mutable objects
print(y)
```

## Part 3: Mutation vs Rebinding

```python
x = [1, 2, 3]
y = x
x.append(4)

x
y
print(f"{id(x)=}")
print(f"{id(y)=}")
x is y
```

```python
x = [1, 2, 3]
y = x
x = [1, 2, 3, 4]

x
y
print(f"{id(x)=}")
print(f"{id(y)=}")
x is y
```

The difference between `x.append(4)` and `x = [1, 2, 3, 4]` is that `x.append(4)` mutates the original object by adding a 4 to it while `x = [1, 2, 3, 4]` creates a new list for x to point to. The append mutates the list in-place while the `x=` creates a new list.

```python
a = 10
b = 3.14
c = "hello"
d = (1, 2, 3)
e = [1, 2, 3]
f = {"a": 1}
g = {1, 2, 3}
```

| Object  | Type  | mutable?  |
|---|---|---|
| a  | int  |  no |
| b |  float |  no |
| c  |  string |  no |
| d  |  tuple |  no |
| e  |  list | yes  |
| f  | dict  | yes  |
| g  |  set |  yes |


```python

print(f"{e=}\t {id(e)=}")
e.append(4)
print(f"{e=}\t {id(e)=}\n")

print(f"{f=}\t {id(f)=}")
f["b"] = 2
print(f"{f=}\t {id(f)=}\n")

print(f"{g=}\t {id(g)=}")
g.add(9)
print(f"{g=}\t {id(g)=}\n")
```

All these objects are modified at the same id (address space), so no new object is created.

### Part 5 — A function that changes its input

```python
def add_value(values):
    values.append(100)

data = [1, 2, 3]
add_value(data)

print(data)
```

```python
def add_value(values):
    result = values.copy()
    result.append(100)
    return result

data = [1, 2, 3]
new_data = add_value(data)

print(data)
print(new_data)
```

**Which style would you prefer for a statistical function, and why?**

Depending on the dataset and how it is cached, I might want to preserve the original data by creating a new variable holding a modified version. If the data is large and the raw data is already saved in it s own location, mutating in place could save space on the computer.

**Explain why a set can be useful for unique categories and membership testing.**

A set only contains one instance of each element with no repeats. This can be helpful for finding the number of unique values by taking the length of a set or a faster way of memebership testing because the code would not need to check redundant values. Each value would only need to be checked once for membership.

**Explain why Python can be dynamically typed while still having types.**

Python infers data types and can cast between types. It is dynamically because types are determined at runtime. Even though the programmer does not need to specify types, Python still uses them - they are just not checked until the code is running.

### Part 13 - Type annotations

```python
def mean(values: list[float]) -> float:
    return sum(values) / len(values)
```

```python
mean([1.0, 2.0, 3.0])
```

```python
mean.__annotations__
```

**Explain what the annotations communicate and why annotations do not make Python statically typed.**

Annotations tell python what type to expect to see for values at runtime, but are not enforced and don't change how the types are processed at execution.

### Part 14 — List comprehensions

```python
squares = []

for x in range(10):
    squares.append(x ** 2)

print(squares)

squares = [x ** 2 for x in range(10)]
print(squares)
```

```python
even_squares = [
    x ** 2
    for x in range(21)
    if x % 2 == 0
]
print(even_squares)
```

```python
# More clearly states the sequence the code itself takes but reads less like a sentence
even_squares = []

for x in range(21):
    if x % 2 == 0:
        even_squares.append(x ** 2)

even_squares
```

### Part 17 — Structured observations

```python
observations = [
    {"id": 1, "score": 82},
    {"id": 2, "score": 91},
    {"id": 3, "score": 76},
    {"id": 4, "score": 88},
    {"id": 5, "score": 94},
]
```

```python
scores = [obs["score"] for obs in observations]
scores
```

```python
high_scores = [
    obs["score"]
    for obs in observations
    if obs["score"] >= 85
]

high_scores
```

```python
score_by_id = {
    obs["id"]: obs["score"]
    for obs in observations
}

score_by_id # return is a dict type, use ID to key score
```

```python
type(observations) # observations is a list of dictionaries with id and score as keys
```

```python
high_scores_by_id = {
    obs["id"]:obs["score"]
    for obs in observations
    if obs["score"] >= 85
}
high_scores_by_id
```

### Part 18 — When not to use a comprehension

A comprehension should not be used when it obscures how the code runs. Writing more complex looking code in fewer lines is not preferrable to cleaner, clearer code that takes up a little more space.
