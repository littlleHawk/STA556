JSON files are hierarchical (nested data). It maps well to a python dictionary. When we read these, we want to flatten it out and loose the hierarchy.

```python
# JSON
import json
```

```python
with open(DATA_DIR/"participants.json") as f:
    records = json.load(f)

type(records) #list of dictionaries

df_json = pd.DataFrame(records)
df_json.head()
```

From an API (Application Programming Interface):
* Most common return from an API is a nested JSON file
* Note: often need an API key to access the data -> allows a person to access a certain amount of data from a call to an API to get data
* keys keep track of who and how often are making calls

```python
with open(DATA_DIR/"api_response.json") as f:
    api_data = json.load(f)

print(api_data.keys())

df_api = pd.DataFrame(api_data["results"]) # Location and measurements will have dictionaries rather than regular columns -> use Normalize
df_api
```

```python
df_api = pd.json_normalize(api_data["results"])
df_api.head()
```
