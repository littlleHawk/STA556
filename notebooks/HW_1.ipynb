```python
non_trad_students["above_avg_gpa"] = "No"
non_trad_students.loc[
    non_trad_students["gpa"] > students["gpa"].mean(),
    "above_avg_gpa"] = "Yes"

non_trad_students
```

I used list comprehension to create the derived variable `above_avg_gpa`, which indicates whether or not the student's GPA is above or below the average of all the students in the dataset. This could be modified to reflect only the students in the subset as well.

# Part 3: Joining Tables

Combine the three tables into one analysis dataset.

Before each merge:
1. state the expected relationship;
2. use an appropriate how=;
3. use validate=;
4. inspect unmatched rows with indicator=True at least once.

Explain any duplicated rows created by a one-to-many relationship.

```python
# Merge Enrollments and Students using a Left Join (we only want courses students are enrolled in)
enrolled_students = enrollments.merge(
    students,
    how = "left",
    left_on = "student_id",
    right_on = "student_id",
    validate = "m:1", # Many enrollments to one student
)

enrolled_students.isna().sum()
```

```python
print(f"Number of Duplicates: {enrolled_students.duplicated().sum()}")
enrolled_students.head()
```

```python
# Merge the enrolled_students with courses, again with a Left join, students may attend multiple courses
final_df = enrolled_students.merge(
    courses,
    how = "left",
    left_on = "course_id",
    right_on = "course_id",
    validate = "m:1", # Many enrollments to one course
    indicator=True
)

print(f"NAs: \n{final_df.isna().sum()}")
print(f"Number of Duplicates: {final_df.duplicated().sum()}")
final_df.head()
```

```python
print(f"students IDs Missing from final_df: {set(students["student_id"]) - set(final_df["student_id"])}\n")
print(f"courses IDs Missing from final_df: {set(courses["course_id"]) - set(final_df["course_id"])}\n")
print(f"enrollments IDs Missing from final_df: {set(enrollments["enrollment_id"]) - set(final_df["enrollment_id"])}\n")
```

```python
final_df.loc[final_df["_merge"] != "both"]
```

There are duplicate students and duplicate courses because one student can be enrolled in multiple courses, and courses can have multiple students. The many - to - one relationships used here indicate the many enrollments for each student and course.
